### Data Preparation

```python
# If there are missing values:
print(df.dtypes)
df.isna().sum() #For caterogiral features, replace them with 'NA' and For numerical features, replace with with 0.0
```

```python
# Fill object (string) columns with "NA"
df[df.select_dtypes(include="object").columns] = (
    df.select_dtypes(include="object").fillna("NA")
)

# Fill numeric columns with 0
df[df.select_dtypes(exclude="object").columns] = (
    df.select_dtypes(exclude="object").fillna(0)
)
df.isna().sum()
```
