Question 3. How many fuel types are presented in the dataset?

Question 4. How many columns in the dataset have missing values?

```python
missing = df.isna().sum()
x = missing[missing != 0].shape[0]
print(f'We have {x} columns with missing values')
```

Question 5. What is the maximum fuel efficiency of cars in Asia?

```python
q = df[df["origin"] == "Asia"]["fuel_efficiency_mpg"].max()
print(f'The fuel efficiency is {q}')
```
